```python
import pandas as pd

df = pd.read_csv('../data/online_retail_raw.csv.csv', encoding='ISO-8859-1')

# How many cancelled orders are there?
cancelled = df[df['InvoiceNo'].astype(str).str.startswith('C')]
print(f"Cancelled transactions: {len(cancelled)}")
print(f"Total transactions: {len(df)}")
```

```python
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])
df['InvoiceDate'].dtype
```

```python
df_clean = df[~df['InvoiceNo'].astype(str).str.startswith('C')]  # remove cancellations
df_clean = df_clean[df_clean['Quantity'] > 0]                     # remove negative/zero quantity
df_clean = df_clean[df_clean['UnitPrice'] > 0]                    # remove free/negative price
df_clean = df_clean.dropna(subset=['CustomerID'])                 # need CustomerID for customer analysis

print(f"Rows before cleaning: {len(df)}")
print(f"Rows after cleaning: {len(df_clean)}")
```

```python
df_clean['TotalPrice'] = df_clean['Quantity'] * df_clean['UnitPrice']
df_clean[['Quantity', 'UnitPrice', 'TotalPrice']].head()
```

```python
print(f"Rows before cleaning: {len(df)}")
print(f"Rows after cleaning: {len(df_clean)}")
print(f"Cancelled transactions removed: {len(cancelled)}")
```

```python
df_clean['InvoiceDate'] = pd.to_datetime(df_clean['InvoiceDate'])
print(df_clean['InvoiceDate'].min())
print(df_clean['InvoiceDate'].max())
```

```python
df_clean.to_csv('../data/online_retail_cleaned.csv', index=False)
print("Saved cleaned data:", df_clean.shape)
```
